```python
def vacuum_cleaner(rooms, pos):
    cost = 0

    while "dirty" in [room.lower() for room in rooms]:

        print("\nCurrent State:", rooms)
        print("Vacuum Location: Room", pos + 1)

        if rooms[pos].lower() == "dirty":
            print("Action: Suck Dirt")
            rooms[pos] = "Clean"
            cost += 1

        else:
            if pos < len(rooms) - 1:
                print("Action: Move Right")
                pos += 1
                cost += 1
            elif pos > 0:
                print("Action: Move Left")
                pos -= 1
                cost += 1

    print("\nFinal State:", rooms)
    print("All rooms cleaned!")
    print("Total Cost:", cost)

print("1BM24CS147")
n = int(input("Enter number of rooms: "))
rooms = []

for i in range(n):
    rooms.append(input(f"Enter status of Room {i+1} (Clean/Dirty): "))

start = int(input("Enter starting room (1 to n): ")) - 1

vacuum_cleaner(rooms, start)d
```
