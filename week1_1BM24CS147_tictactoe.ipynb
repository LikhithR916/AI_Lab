### Tic Tac Toe Game
This interactive notebook implements a classic two-player Tic Tac Toe game. Players will take turns entering their moves.

```python

def print_board(board):
    print(f" {board[0]} | {board[1]} | {board[2]} ")
    print("---|---|---")
    print(f" {board[3]} | {board[4]} | {board[5]} ")
    print("---|---|---")
    print(f" {board[6]} | {board[7]} | {board[8]} ")

def check_winner(board):
    win_conditions = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8), # Rows
        (0, 3, 6), (1, 4, 7), (2, 5, 8), # Columns
        (0, 4, 8), (2, 4, 6)             # Diagonals
    ]
    for combo in win_conditions:
        if board[combo[0]] == board[combo[1]] == board[combo[2]] != ' ':
            return board[combo[0]]
    if ' ' not in board:
        return 'Draw'
    return None

def play_game():
    board = [' '] * 9
    current_player = 'X'
    player_names = {'X': 'Player 1 (X)', 'O': 'Player 2 (O)'}

    print("Welcome to Tic Tac Toe!")
    print("Positions are mapped from 1 to 9 as follows:")
    print_board([str(i) for i in range(1, 10)])
    print("\nGame starts!\n")

    while True:
        print_board(board)
        try:
            move = int(input(f"{player_names[current_player]}, enter your move (1-9): ")) - 1
            if move < 0 or move > 8 or board[move] != ' ':
                print("Invalid move! Please try again.")
                continue
        except ValueError:
            print("Please enter a valid number between 1 and 9.")
            continue

        board[move] = current_player
        result = check_winner(board)

        if result:
            print_board(board)
            if result == 'Draw':
                print("It's a draw!")
            else:
                print(f"Congratulations! {player_names[result]} wins!")
            break

        current_player = 'O' if current_player == 'X' else 'X'

# Run the game
print("1BM24CS147_LikhithR")
play_game()
```
