```python
def is_variable(expr):
    """
    Helper function to determine if an expression is a variable.
    By convention, single lowercase letters (a-z) are treated as variables,
    except for 'a', 'b', 'c' which are often constants. Let's define
    variables as strings starting with u, v, w, x, y, z.
    """
    if isinstance(expr, str):
        return expr.strip() in ['x', 'y', 'z', 'u', 'v', 'w']
    return False

def is_constant(expr):
    """
    Helper function to determine if an expression is a constant.
    Single letters like a, b, c or integers are constants.
    """
    if isinstance(expr, str):
        return expr.strip() in ['a', 'b', 'c', 'd'] or expr.isdigit()
    return False

def occurs_in(var, expr):
    """
    Checks if a variable 'var' occurs inside another expression 'expr' (Occurs Check).
    """
    if var == expr:
        return True
    if isinstance(expr, list):
        # Search through predicate/function arguments
        return any(occurs_in(var, arg) for arg in expr[1:])
    return False

def apply_substitution(subst, expr):
    """
    Applies a substitution set (dictionary or list of tuples) to an expression.
    """
    if not subst or subst == "NIL":
        return expr

    # If expr is a variable, check if it has a substitution
    if isinstance(expr, str):
        for s_val, s_var in subst:
            if expr == s_var:
                return apply_substitution(subst, s_val)
        return expr

    # If expr is a list (predicate or function), apply recursively to arguments
    if isinstance(expr, list):
        return [expr[0]] + [apply_substitution(subst, arg) for arg in expr[1:]]
    return expr

def unify(psi1, psi2):
    # Step 1: If Psi1 or Psi2 is a variable or constant, then:
    if (is_variable(psi1) or is_constant(psi1)) or (is_variable(psi2) or is_constant(psi2)):
        # a) If Psi1 or Psi2 are identical, then return NIL
        if psi1 == psi2:
            return "NIL"

        # b) Else if Psi1 is a variable
        elif is_variable(psi1):
            # a. then if Psi1 occurs in Psi2, then return FAILURE
            if occurs_in(psi1, psi2):
                return "FAILURE"
            # b. Else return {(Psi2 / Psi1)}
            else:
                return [(psi2, psi1)]

        # c) Else if Psi2 is a variable
        elif is_variable(psi2):
            # a. If Psi2 occurs in Psi1 then return FAILURE
            if occurs_in(psi2, psi1):
                return "FAILURE"
            # b. Else return {(Psi1 / Psi2)}
            else:
                return [(psi1, psi2)]

        # d) Else return FAILURE
        else:
            return "FAILURE"

    # Step 2: If the initial Predicate symbol in Psi1 and Psi2 are not same, then return FAILURE
    if psi1[0] != psi2[0]:
        return "FAILURE"

    # Step 3: If Psi1 and Psi2 have a different number of arguments, then return FAILURE
    if len(psi1) != len(psi2):
        return "FAILURE"

    # Step 4: Set Substitution set(SUBST) to NIL
    subst = "NIL"

    # Step 5: For i=1 to the number of elements (arguments) in Psi1
    # Note: index 0 is the predicate name, so we iterate through arguments from index 1 onwards
    for i in range(1, len(psi1)):
        # a) Call Unify function with the ith element of Psi1 and ith element of Psi2, and put the result into S
        s = unify(psi1[i], psi2[i])

        # b) If S = failure then returns Failure
        if s == "FAILURE":
            return "FAILURE"

        # c) If S != NIL then do
        if s != "NIL":
            # a. Apply S to the remainder of both L1 and L2
            # Dynamically update the rest of the arguments with the new substitution
            for j in range(i + 1, len(psi1)):
                psi1[j] = apply_substitution(s, psi1[j])
                psi2[j] = apply_substitution(s, psi2[j])

            # b. SUBST = APPEND(S, SUBST)
            if subst == "NIL":
                subst = []
            subst = s + subst

    # Step 6: Return SUBST
    return subst

# --- User Input & Execution Section ---
if __name__ == "__main__":
    print("--- First-Order Logic Unification Algorithm ---")
    print("Variables convention: x, y, z, u, v, w")
    print("Constants convention: a, b, c, d")
    print("Format expressions as Python lists. Example: ['P', 'x', ['f', 'y']]\n")

    try:
        # Taking inputs dynamically from the user
        input1 = input("Enter first expression (Psi1): ")
        input2 = input("Enter second expression (Psi2): ")

        # Safely evaluate string input into Python list structures
        psi1 = eval(input1)
        psi2 = eval(input2)

        print(f"\nUnifying:\nPsi1 = {psi1}\nPsi2 = {psi2}")

        result = unify(psi1, psi2)

        print("\n--- Result ---")
        if result == "FAILURE" or result == "NIL":
            print(result)
        else:
            # Formatting output substitution set: {val/var}
            formatted_subst = ", ".join([f"{{{pair[0]}/{pair[1]}}}" for pair in result])
            print(f"SUBST = [ {formatted_subst} ]")

    except Exception as e:
        print(f"\nError parsing input: {e}")
        print("Make sure to type valid python list notation, e.g., ['P', 'x', ['f', 'y']]")
```
