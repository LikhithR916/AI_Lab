```python
import math


def alpha_beta(depth, node_idx, is_max, values, alpha, beta, bf, pruned):
    # Base case: reached a leaf node
    if depth == 0:
        return values[node_idx], [f"Leaf[{node_idx}]={values[node_idx]}"]

    if is_max:
        best_val = -math.inf
        best_path = []

        for i in range(bf):
            child_idx = node_idx * bf + i
            val, path = alpha_beta(
                depth - 1, child_idx, False, values, alpha, beta, bf, pruned
            )

            if val > best_val:
                best_val = val
                best_path = [f"MAX({best_val})"] + path

            alpha = max(alpha, best_val)

            # Beta cutoff
            if alpha >= beta:
                # Record skipped leaf range
                skipped_start = (node_idx * bf + i + 1) * (bf ** (depth - 1))
                skipped_end = (node_idx * bf + bf) * (bf ** (depth - 1))
                pruned.append(
                    f"Pruned values {values[skipped_start:skipped_end]} (α={alpha}"
                    f" >= β={beta})"
                )
                break

        return best_val, best_path

    else:
        best_val = math.inf
        best_path = []

        for i in range(bf):
            child_idx = node_idx * bf + i
            val, path = alpha_beta(
                depth - 1, child_idx, True, values, alpha, beta, bf, pruned
            )

            if val < best_val:
                best_val = val
                best_path = [f"MIN({best_val})"] + path

            beta = min(beta, best_val)

            # Alpha cutoff
            if alpha >= beta:
                skipped_start = (node_idx * bf + i + 1) * (bf ** (depth - 1))
                skipped_end = (node_idx * bf + bf) * (bf ** (depth - 1))
                pruned.append(
                    f"Pruned values {values[skipped_start:skipped_end]} (α={alpha}"
                    f" >= β={beta})"
                )
                break

        return best_val, best_path


# --- Simple 2-Line User Input ---
bf = int(input("Enter branching factor (e.g., 2 for binary tree): "))
leaves = list(map(int, input("Enter all leaf values separated by space: ").split()))

# Automatically calculate tree depth
depth = round(math.log(len(leaves), bf))

pruned_list = []
root_value, optimal_path = alpha_beta(
    depth, 0, True, leaves, -math.inf, math.inf, bf, pruned_list
)

print("\nRoot Node Value :", root_value)
print("Optimal Path    :", " -> ".join(optimal_path))
print("Pruned Branches :")
for p in pruned_list:
    print(" -", p)
```
