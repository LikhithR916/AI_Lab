```python

import random
import math

N = 8

def heuristic(board):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing():
    while True:
        board = list(range(N))
        random.shuffle(board)

        temperature = 100.0
        cooling_rate = 0.995

        while temperature > 0.01:
            current_cost = heuristic(board)

            if current_cost == 0:
                return board

            new_board = board.copy()

            # Swap two queens to create a neighboring state
            i, j = random.sample(range(N), 2)
            new_board[i], new_board[j] = new_board[j], new_board[i]

            new_cost = heuristic(new_board)
            delta = new_cost - current_cost

            # Accept a better move or sometimes a worse move
            if delta < 0 or random.random() < math.exp(-delta / temperature):
                board = new_board

            temperature *= cooling_rate

# Find a solution
solution = simulated_annealing()
print("Likhith R - 1BM24CS147")
print("Solution found!")
print("Board representation:", solution)
print("\nChessboard:")

for row in range(N):
    for col in range(N):
        if solution[col] == row:
            print("Q", end=" ")
        else:
            print(".", end=" ")
    print()

print("\nNumber of conflicts:", heuristic(solution))
```
