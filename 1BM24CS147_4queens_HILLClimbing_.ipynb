```python
def calculate_cost(state):
    """Calculates h(n): the number of mutually attacking pairs of queens."""
    attacks = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            # Checking only diagonal conflicts as rows/columns are distinct permutations
            if abs(state[i] - state[j]) == abs(i - j):
                attacks += 1
    return attacks

def print_board(state, label="Board Layout"):
    """Prints a clear 4x4 visual text layout of the board configuration."""
    print(f"--- {label} ---")
    for row in range(4):
        print(" ".join("Q" if state[col] == row else "." for col in range(4)))
    print("-" * (len(label) + 8))

def get_matrix_input():
    """Prompts row-by-row for the initial matrix layout."""
    print("Enter 4x4 matrix row-by-row (use 1 for Queen, 0 for empty):")
    queen_positions = [-1] * 4
    for r in range(4):
        while True:
            try:
                row = [int(x) for x in input(f"Row {r+1}: ").split()]
                if len(row) != 4:
                    print("Error: Must enter exactly 4 values.")
                    continue
                for c, val in enumerate(row):
                    if val == 1:
                        queen_positions[c] = r
                break
            except ValueError:
                print("Error: Invalid integers.")
    return tuple(queen_positions)

def continuous_search(initial_state):
    """Explores neighbors using pair-swaps until cost h(n) = 0."""
    current = initial_state
    cost = calculate_cost(current)
    visited = {current}

    print("\n==============================")
    print_board(current, "INITIAL BOARD STATE")
    print("==============================\n")

    step = 1
    while True:
        state_str = str(list(current)).replace(" ", "")
        print(f"\nNeighbours of {state_str} (cost is {cost}.)")

        if cost == 0:
            print("\n==============================")
            print_board(current, "FINAL GOAL STATE")
            print("*** SUCCESS: Perfect solution found! ***")
            print("==============================")
            break

        all_possible_neighbors = []

        # Generate all 6 unique position swap options
        for i in range(4):
            for j in range(i + 1, 4):
                neighbor = list(current)
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]

                n_state = tuple(neighbor)
                n_cost = calculate_cost(n_state)
                all_possible_neighbors.append((n_cost, n_state))

        # Identify the best candidate unvisited option
        best_unvisited_cost = float('inf')
        best_unvisited_state = None

        for n_cost, n_state in all_possible_neighbors:
            if n_state not in visited and n_cost < best_unvisited_cost:
                best_unvisited_cost = n_cost
                best_unvisited_state = n_state

        # Print all choices matching requested text structure
        for n_cost, n_state in all_possible_neighbors:
            n_str = str(list(n_state)).replace(" ", "")
            seen_flag = " (Visited)" if n_state in visited else ""

            if n_state == best_unvisited_state:
                print(f" • {n_str} cost is {n_cost}. Choose this neighbour.")
            else:
                print(f" • {n_str} cost is {n_cost}.{seen_flag}")

        if best_unvisited_state is None:
            print("\nDead end. No unvisited neighbors remaining.")
            break

        # Log transition path to intermediate state
        next_str = str(list(best_unvisited_state)).replace(" ", "")
        print(f"\n>>> Transition: {state_str} -> {next_str}")
        print_board(best_unvisited_state, f"INTERMEDIATE STATE STEP {step}")

        # Update pointers
        current = best_unvisited_state
        cost = best_unvisited_cost
        visited.add(current)
        step += 1

if __name__ == "__main__":
    start_state = get_matrix_input()
    continuous_search(start_state)
```
